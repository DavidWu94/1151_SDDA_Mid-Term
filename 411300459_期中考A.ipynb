````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

def init_db():
    conn = sqlite3.connect('association.db')
    cursor = conn.cursor()

    try:
        # 刪除舊資料表
        cursor.execute("DROP TABLE IF EXISTS registrations")
        cursor.execute("DROP TABLE IF EXISTS activities")

        # 建立 activities 資料表
        cursor.execute("""
            CREATE TABLE activities (
                activity_name TEXT PRIMARY KEY,
                activity_date TEXT NOT NULL,
                capacity INTEGER NOT NULL
            )
        """)

        # 建立 registrations 資料表
        cursor.execute("""
            CREATE TABLE registrations (
                reg_id INTEGER PRIMARY KEY AUTOINCREMENT,
                student_name TEXT NOT NULL,
                activity_name TEXT NOT NULL,
                num_people INTEGER NOT NULL DEFAULT 1,
                status TEXT NOT NULL DEFAULT '已報名',
                FOREIGN KEY (activity_name) REFERENCES activities (activity_name)
            )
        """)

        # 插入初始活動資料
        initial_activities = [
            ('迎新晚會', '2026-11-20', 6),
            ('程式工作坊', '2026-11-25', 4),
            ('籃球友誼賽', '2026-11-28', 3)
        ]

        cursor.executemany(
            "INSERT INTO activities (activity_name, activity_date, capacity) VALUES (?, ?, ?)",
            initial_activities
        )

        conn.commit()
        print("✅ 資料庫與初始活動資料建立成功！")
    except Exception as e:
        conn.rollback()
        print(f"❌ 建立資料庫時發生錯誤: {e}")
        raise e
    finally:
        conn.close()

# 執行資料庫初始化
init_db()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

def show_tables():
    conn = sqlite3.connect('association.db')
    try:
        print("--- activities 資料表 ---")
        df_activities = pd.read_sql_query("SELECT * FROM activities", conn)
        display(df_activities)

        print("\n--- registrations 資料表 ---")
        df_regs = pd.read_sql_query("SELECT * FROM registrations", conn)
        display(df_regs)
    except Exception as e:
        print(f"❌ 讀取資料表時發生錯誤: {e}")
    finally:
        conn.close()

# 呼叫一次顯示目前資料
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

# 建立快取，避免對同一句話重複發送 API 請求
ai_cache = {}

def parse_message_with_ai(message):
    """
    使用 Groq API 來理解使用者輸入的一句話，並解析為 (動作, 活動, 人數, 原始回覆) 的格式。
    """
    # 先檢查快取
    if message in ai_cache:
        return ai_cache[message]

    # 定義允許清單與格式防禦
    allowed_actions = ['報名', '取消', '查詢', '統計']
    allowed_activities = ['迎新晚會', '程式工作坊', '籃球友誼賽']

    system_prompt = """
    你是一個系學會報名系統的語意解析助手。
    請根據使用者輸入的一句話，解析出：動作、活動、以及人數。

    輸出規範：
    1. 請嚴格只輸出「動作,活動,人數」這三個欄位，中間以半形逗號分隔，不要含有其他任何文字、空格、標點符號或說明。
    2. 動作只能是：報名、取消、查詢、統計、未知 其中的一個。
       特別注意：任何詢問「各活動報名狀況」、「報名統計」、「活動狀況」等整體性報表的句子，其動作必須判定為「統計」，活動欄位填「未知」。
       而詢問特定單一活動剩幾人（如：程式工作坊還有多少位置）的，其動作判定為「查詢」。
    3. 活動只能是：迎新晚會、程式工作坊、籃球友誼賽、未知 其中的一個。
    4. 人數：請根據句子內提到的總人數（例如「我和兩個朋友」代表3人），若沒提到任何數字，預設為 1 人。
    5. 如果無法明確判斷動作或活動，請對應欄位回傳 '未知'。例如：「未知,未知,1」
    """

    raw_response = ""
    retries = 3
    for i in range(retries):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[
                    {"role": "system", "content": system_prompt},
                    {"role": "user", "content": message}
                ],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            raw_response = response.choices[0].message.content.strip()
            time.sleep(2) # 遵守間隔限制
            break
        except Exception as e:
            # 偵測 429 或是 API 錯誤時等待並重試
            if "429" in str(e) and i < retries - 1:
                print(f"⚠️ 遭遇 API 流量限制或錯誤，等待 30 秒後重試... ({i+1}/{retries})")
                time.sleep(30)
            else:
                print(f"❌ 呼叫 AI 發生錯誤: {e}")
                raw_response = "未知,未知,1"
                break

    # 解析與安全性防禦
    action = "未知"
    activity = "未知"
    people = 1

    # 切割 CSV 格式
    parts = [p.strip() for p in raw_response.split(',')]
    if len(parts) >= 3:
        temp_action, temp_activity, temp_people_str = parts[0], parts[1], parts[2]

        # 1. 動作篩選
        if temp_action in allowed_actions:
            action = temp_action

        # 2. 活動篩選
        if temp_activity in allowed_activities:
            activity = temp_activity

        # 3. 正則表達式抓取人數整數
        nums = re.findall(r'\d+', temp_people_str)
        if nums:
            people = int(nums[0])
        else:
            people = 1

    result = (action, activity, people, raw_response)
    # 存入快取
    ai_cache[message] = result
    return result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

def run_ai_tests():
    # 測試集格式: (使用者輸入的一句話, 預期動作, 預期活動, 預期人數)
    # 若預期對象為 '*' 代表為未知（或不在意特定值）
    test_cases = [
        ("我要報名迎新晚會", "報名", "迎新晚會", 1),
        ("幫我和兩個朋友報名程式工作坊", "報名", "程式工作坊", 3),
        ("取消籃球友誼賽", "取消", "籃球友誼賽", 1),
        ("請問程式工作坊還有多少位置？", "查詢", "程式工作坊", 1),
        ("顯示各活動報名狀況", "統計", "未知", 1),
        ("我想幫五個人報名迎新晚會", "報名", "迎新晚會", 5),
        ("我要報名羽球比賽", "報名", "未知", 1),
        ("亂七八糟看不懂的話", "未知", "未知", 1)
    ]

    passed_count = 0

    print(f"======= 開始 AI 解讀測試 (共 {len(test_cases)} 題) =======\n")

    for idx, (msg, exp_action, exp_activity, exp_people) in enumerate(test_cases, 1):
        action, activity, people, raw = parse_message_with_ai(msg)

        # 進行比對驗證
        match_action = (action == exp_action)
        match_activity = (activity == exp_activity)
        match_people = (people == exp_people)

        is_passed = match_action and match_activity and match_people

        if is_passed:
            passed_count += 1
            status_str = "✅ 通過"
        else:
            status_str = "❌ 失敗"

        print(f"測試 #{idx}: \"{msg}\"")
        print(f"  ├ 預期結果: [{exp_action}, {exp_activity}, {exp_people}]")
        print(f"  ├ 系統解讀: [{action}, {activity}, {people}]")
        print(f"  ├ AI 原始回覆: \"{raw}\"")
        print(f"  └ 測試結果: {status_str}\n")

    print(f"======= 測試結束 =======")
    print(f"🎯 通過題數: {passed_count} / {len(test_cases)}")

# 執行測試
run_ai_tests()
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

def process_system_flow(student_name, message):
    """
    處理使用者的一句話：解析、判定、並對 SQLite 資料庫執行對應的操作。
    """
    # 先使用 AI 解析
    action, activity, people, raw_response = parse_message_with_ai(message)

    # 例外處理：若無法判斷活動，回覆清單
    if action in ['報名', '取消', '查詢'] and activity == '未知':
        return "❌ 無法判斷活動，目前可報名的活動有：迎新晚會、程式工作坊、籃球友誼賽。"

    if action == '未知':
        return "❌ 無法識別您的意圖，請說出類似「我要報名迎新晚會」或「查詢程式工作坊名額」的句子。"

    conn = sqlite3.connect('association.db')
    cursor = conn.cursor()

    try:
        # === 1. 報名邏輯 ===
        if action == '報名':
            if not student_name or student_name.strip() == "":
                return "❌ 報名失敗：報名需要填寫您的姓名。"

            # 檢查單次報名人數規則 (1~3人)
            if people < 1 or people > 3:
                return f"❌ 報名失敗：單次報名人數必須在 1 至 3 人之間。(您輸入的人數：{people}人)"

            # A2 每人上限規則：檢查該學生目前已報名的有效活動數量(排除已取消)
            cursor.execute("""
                SELECT COUNT(DISTINCT activity_name)
                FROM registrations
                WHERE student_name = ? AND status = '已報名'
            """, (student_name,))
            count_row = cursor.fetchone()
            current_active_count = count_row[0] if count_row is not None else 0

            if current_active_count >= 2:
                return "❌ 報名失敗：每人最多只能同時報名 2 個活動！"

            # 檢查同一人是否重複報名同一活動
            cursor.execute("""
                SELECT COUNT(*) FROM registrations
                WHERE student_name = ? AND activity_name = ? AND status = '已報名'
            """, (student_name, activity))
            dup_row = cursor.fetchone()
            if dup_row is not None and dup_row[0] > 0:
                return f"❌ 報名失敗：您已報名過此活動「{activity}」！"

            # 查詢活動總容量
            cursor.execute("SELECT capacity FROM activities WHERE activity_name = ?", (activity,))
            act_row = cursor.fetchone()
            if act_row is None:
                return f"❌ 報名失敗：查無活動「{activity}」。"
            capacity = act_row[0]

            # 查詢已被佔用的名額加總 (以人數加總計算)
            cursor.execute("""
                SELECT SUM(num_people) FROM registrations
                WHERE activity_name = ? AND status = '已報名'
            """, (activity,))
            sum_row = cursor.fetchone()
            occupied = sum_row[0] if (sum_row is not None and sum_row[0] is not None) else 0

            # 剩餘名額
            remaining = capacity - occupied
            if people > remaining:
                return f"❌ 報名失敗：剩餘名額不足。目前僅剩 {remaining} 個位置，而您欲報名 {people} 人。"

            # 執行報名
            cursor.execute("""
                INSERT INTO registrations (student_name, activity_name, num_people, status)
                VALUES (?, ?, ?, '已報名')
            """, (student_name, activity, people))
            conn.commit()
            return f"🎉 報名成功！{student_name} 成功報名「{activity}」共 {people} 人。"

        # === 2. 取消邏輯 ===
        elif action == '取消':
            if not student_name or student_name.strip() == "":
                return "❌ 取消失敗：取消需要填寫您的姓名。"

            # 查詢是否有已報名紀錄
            cursor.execute("""
                SELECT reg_id FROM registrations
                WHERE student_name = ? AND activity_name = ? AND status = '已報名'
            """, (student_name, activity))
            reg_row = cursor.fetchone()
            if reg_row is None:
                return f"❌ 取消失敗：找不到您在「{activity}」的有效報名紀錄。"

            reg_id = reg_row[0]
            # 更新狀態為已取消
            cursor.execute("""
                UPDATE registrations
                SET status = '已取消'
                WHERE reg_id = ?
            """, (reg_id,))
            conn.commit()
            return f"✅ 取消成功！已成功取消您在「{activity}」的報名。"

        # === 3. 查詢名額 ===
        elif action == '查詢':
            # 查詢總名額
            cursor.execute("SELECT capacity FROM activities WHERE activity_name = ?", (activity,))
            act_row = cursor.fetchone()
            if act_row is None:
                return f"❌ 查詢失敗：找不到活動「{activity}」。"
            capacity = act_row[0]

            # 查詢已佔用人數
            cursor.execute("""
                SELECT SUM(num_people) FROM registrations
                WHERE activity_name = ? AND status = '已報名'
            """, (activity,))
            sum_row = cursor.fetchone()
            occupied = sum_row[0] if (sum_row is not None and sum_row[0] is not None) else 0

            remaining = capacity - occupied
            return f"ℹ️ 剩餘名額查詢：「{activity}」剩餘名額：{remaining} 人。"

        # === 4. 統計查詢 (A3) ===
        elif action == '統計':
            # 使用 LEFT JOIN 與 GROUP BY，確保沒有人報名的活動也會被列出
            cursor.execute("""
                SELECT
                    a.activity_name,
                    a.activity_date,
                    a.capacity,
                    COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.num_people ELSE 0 END), 0) as registered_count
                FROM activities a
                LEFT JOIN registrations r ON a.activity_name = r.activity_name
                GROUP BY a.activity_name
            """)
            rows = cursor.fetchall()

            report_lines = ["📊 【各活動報名統計報表】"]
            for row in rows:
                act_name, act_date, cap, reg_cnt = row[0], row[1], row[2], row[3]
                rem = cap - reg_cnt
                report_lines.append(f"- 📍 {act_name} ({act_date}) | 總名額：{cap} 人 | 已報名：{reg_cnt} 人 | 剩餘：{rem} 人")
            return "\n".join(report_lines)

    except Exception as e:
        conn.rollback()
        return f"❌ 系統處理程序發生錯誤: {e}"
    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

def run_flow_tests():
    # 先將資料庫與初始資料還原，確保測試起點乾淨
    print("🔄 正在初始化/還原資料庫...")
    init_db()
    print("\n======= 開始 7-2 流程測試 (共 9 步) =======\n")

    flow_steps = [
        # (步驟編號, 學生姓名, 語句, 預期說明)
        (1, "張三", "我要報名迎新晚會", "報名成功！"),
        (2, "張三", "幫我和一個朋友報名程式工作坊", "報名成功！(張三已報名 2 個活動)"),
        (3, "張三", "我要報名籃球友誼賽", "報名失敗：每人最多只能報名2個活動"),
        (4, "李四", "幫我和一個朋友報名程式工作坊", "報名成功！(此時程式工作坊剩餘名額變 0)"),
        (5, "王五", "我要報名程式工作坊", "報名失敗：剩餘名額不足"),
        (6, "張三", "取消程式工作坊", "取消成功！(名額釋出 2)"),
        (7, "王五", "我要報名程式工作坊", "報名成功！"),
        (8, "", "請問程式工作坊還有多少位置？", "查詢剩餘名額：1 人"),
        (9, "", "各活動報名狀況", "顯示包含迎新、工作坊、籃球賽的三筆統計報表")
    ]

    for step, name, sentence, expected in flow_steps:
        print(f"🔹 [步驟 {step}] 測試人員: {name if name else '(無)'} | 語句: \"{sentence}\"")
        print(f"   [預期結果] {expected}")
        response = process_system_flow(name, sentence)
        print(f"   [系統回應] {response}\n")

    print("======= 流程測試結束 =======\n")

    # 測試結束後，展示資料庫最終狀況，供評分核對
    print("📊 ======= 最終資料庫資料表展示 =======")
    show_tables()

# 執行流程測試
run_flow_tests()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# 提醒：if IN_COLAB: demo.launch(share=True)
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）

def create_gradio_interface():
    # 定義 Gradio 主要處理函式
    def handle_gradio_input(student_name, message):
        if not message or message.strip() == "":
            return "⚠️ 請輸入您的一句話後再點擊送出。"
        return process_system_flow(student_name, message)

    # 使用 Gradio Blocks 設計簡潔美觀的界面
    with gr.Blocks(title="系學會活動智慧報名系統") as demo:
        gr.Markdown("# 🎓 系學會活動智慧報名系統 (期中考 A 卷)")
        gr.Markdown("直接輸入您的一句話，系統將自動解析並為您辦理活動「報名、取消、查詢名額及統計」！")

        with gr.Row():
            with gr.Column():
                student_name_input = gr.Textbox(
                    label="學生姓名",
                    placeholder="報名或取消時必須填寫姓名，查詢名額與統計則不需...",
                    lines=1
                )
                message_input = gr.Textbox(
                    label="請輸入一句話",
                    placeholder="例如：'我要報名迎新晚會' 或 '幫我和兩個朋友報名程式工作坊' 或 '各活動報名狀況'",
                    lines=2
                )
                submit_btn = gr.Button("🚀 點我送出", variant="primary")

            with gr.Column():
                output_display = gr.Textbox(
                    label="系統回應結果",
                    interactive=False,
                    lines=8
                )

        # 綁定事件
        submit_btn.click(
            fn=handle_gradio_input,
            inputs=[student_name_input, message_input],
            outputs=output_display
        )

    # 依題目規定，若在 Colab 環境則開啟公開分享連結
    if IN_COLAB:
        demo.launch(share=True, debug=False)
    else:
        demo.launch()

# 啟動網頁界面
create_gradio_interface()
````
